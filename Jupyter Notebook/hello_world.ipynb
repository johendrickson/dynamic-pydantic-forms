```python
"""
A simple interactive Jupyter Notebook UI using ipywidgets.

This notebook creates a small form that allows users to enter their name
and click a button to display a greeting. It demonstrates:

- Using `ipywidgets` to create interactive form elements (Text input, Button)
- Using `IPython.display` to render output dynamically in the notebook
- Handling widget events and updating output in response to user interaction
"""
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create widgets
name_input = widgets.Text(
    value="",
    placeholder="Enter your name",
    description="Name:",
)

greet_button = widgets.Button(
    description="Say Hello 👋",
    button_style="success",
)

output = widgets.Output()

# Define interaction
def on_greet_click(b):
    with output:
        clear_output()
        name = name_input.value.strip() or "World"
        print(f"Hello, {name}!")

# Link button click to function
greet_button.on_click(on_greet_click)

# Display UI
display(name_input, greet_button, output)
```
